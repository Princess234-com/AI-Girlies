```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('regional_demand.csv')
```

```python
regional_averages = df.groupby('Region')['ForecastDemand100'].mean()
print(regional_averages)
```

```python
total_demand = regional_averages.sum()
regional_percentages = (regional_averages / total_demand) * 100
print(regional_percentages)
print(total_demand)
```
